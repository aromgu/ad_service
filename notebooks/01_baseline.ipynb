```python
from pathlib import Path

from ad_service.api.schemas.generation import GenerationRequest
from ad_service.factory import (
    create_background_remover,
    create_copy_provider,
    create_image_provider,
)
from ad_service.pipelines.inference import GenerationPipeline
```

```python
request_path = Path('data/processed/eval_v1/requests/snack_001.json')
request = GenerationRequest.model_validate_json(request_path.read_text(encoding='utf-8'))
pipeline = GenerationPipeline(
    copy_provider=create_copy_provider('mock'),
    image_provider=create_image_provider('mock'),
    background_remover=create_background_remover('simple'),
    output_root=Path('data/outputs/colab_mock'),
)
result = pipeline.generate(request, base_dir=Path.cwd())
result
```
